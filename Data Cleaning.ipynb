```python
# ---
# Make a copy before cleaning
df_clean = df.copy()
```

```python
# -----------------------------
# 1. Clean Age
# -----------------------------
# Ages outside a reasonable range are treated as missing
df_clean.loc[(df_clean['Age'] < 18) | (df_clean['Age'] > 100), 'Age'] = np.nan
```

```python
# ---
# Fill invalid/missing ages with median age
df_clean['Age'] = df_clean['Age'].fillna(df_clean['Age'].median())
```

```python
# ---
# Convert Age to integer
df_clean['Age'] = df_clean['Age'].astype(int)
```

```python
# -----------------------------
# 2. Clean Gender  [FIXED]
# -----------------------------
male_set = {'m','male','male-ish','maile','mal','male (cis)','make','man',
            'msle','mail','malr','cis male','cis man','guy (-ish) ^_^',
            'male leaning androgynous'}
female_set = {'f','female','cis female','woman','femake','cis-female/femme',
              'female (cis)','femail'}

def clean_gender(value):
    value = str(value).strip().lower()
    if value in female_set:
        return 'Female'
    elif value in male_set:
        return 'Male'
    else:
        return 'Other'

df_clean['Gender'] = df_clean['Gender'].apply(clean_gender)
```

```python
# -----------------------------
# 3. Convert Timestamp
# -----------------------------
df_clean['Timestamp'] = pd.to_datetime(df_clean['Timestamp'], errors='coerce')
```

```python
# ---
# Check results
print("Age range:", df_clean['Age'].min(), "-", df_clean['Age'].max())
print("\nGender categories:")
print(df_clean['Gender'].value_counts())
print("\nMissing values after basic cleaning:")
print(df_clean.isnull().sum())
```

```python
# ---
# Final missing-value treatment
df_clean['state'] = df_clean['state'].fillna('Not Specified')
df_clean['self_employed'] = df_clean['self_employed'].fillna('Unknown')
df_clean['work_interfere'] = df_clean['work_interfere'].fillna('Not Applicable')
```

```python
# Comments has too many missing values and is not required for our structured EDA
df_clean.drop(columns=['comments'], inplace=True)
```

```python
# Final checks
print("Shape after cleaning:", df_clean.shape)
print("\nRemaining missing values:")
print(df_clean.isnull().sum().sum())
print("\nData types:")
print(df_clean.dtypes)
```

```python
# Duplicate check  [FIXED - new cell]
# NOTE: every row has a unique Timestamp, so a duplicate check that
# includes Timestamp will always show 0. Check on the actual survey
# answers instead.
answer_cols = [c for c in df_clean.columns if c != 'Timestamp']
print("Duplicate rows (all columns incl. Timestamp):", df_clean.duplicated().sum())
print("Duplicate rows (survey answers only):", df_clean.duplicated(subset=answer_cols).sum())

df_clean = df_clean.drop_duplicates(subset=answer_cols).reset_index(drop=True)
print("Shape after removing duplicates:", df_clean.shape)
```
